```python
# Blank Review rating needs to be fixed;
# Mean or Median
# Median > Mean -> Choose bcoz Mean can be affected by outliers; Medium is robust
```

```python
#category-wise rating
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
# we can also use isna() to check if there is any null value
df.isna().sum()
```

```python
# snake casing
# eg. customer_id
df.columns= df.columns.str.lower()
#df.columns
df.columns= df.columns.str.replace(' ','_')
#df.columns
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
# Feature Engineering
# Create New Columns
# Age Group - YoungAdult, Adult, MiddleAge, Senior

labels = ['Young Adults','Adults','Middle-Aged','Senior']
df['age_group']= pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#Purchase Frequency days
df['frequency_of_purchases'].unique()
```

```python
# Creating a dictionary
frequency_mapping={
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_freq_days']= df['frequency_of_purchases'].map(frequency_mapping)
```
